```python
import pandas as pd
from sklearn.ensemble import GradientBoostingRegressor
```

```python
train_df = pd.read_csv(r"C:\potfolio\Advance_house_price_prediction\data\processed_data\processed_train.csv")
train_df.head()
```

```python
missing = {} 
length = len(train_df)
for col in train_df.columns:
    missing_val = train_df[col].isna().sum()
    missing[col] = (missing_val/length) * 100
missing
```

```python
drop_col = []
for key, value in missing.items():
    if value > 0 :
        drop_col.append(key)
drop_col
```

```python
for  col in drop_col:
    train_df[col] = train_df[col].fillna(train_df[col].mean())

train_df.head()
```

```python
train_df.isna().sum()
```

```python
train_df["SalePrice"]
```

```python
from sklearn.model_selection import train_test_split

x = train_df.drop(columns= ["SalePrice"])
y = train_df["SalePrice"]

x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.3, random_state=42)

gd = GradientBoostingRegressor(
    n_estimators=2000,
    learning_rate=0.01,
    max_depth=5,
    subsample=0.8,
    min_samples_split=12,
    random_state=42,
    criterion="squared_error"
)
```

```python
gd.fit(x_train,y_train)
```

```python
y_pred = gd.predict(x_test)


import numpy as np 
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
y_pred_actual = np.expm1(y_pred)
y_test_actual = np.expm1(y_test)

print("RMSE:", np.sqrt(mean_squared_error(y_test_actual, y_pred_actual)))
print("MAE:", mean_absolute_error(y_test_actual, y_pred_actual))
print("R2:", r2_score(y_test_actual, y_pred_actual))
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(gd, x, y, cv=5, scoring="r2")
print("CV R2:", scores.mean())
```

```python
from xgboost import XGBRegressor

xgb = XGBRegressor(
    n_estimators=2500, max_depth=5, subsample=0.7,
    min_samples_split=12,
    random_state=42,learning_rate=0.03
)

xgb.fit(x_train,y_train)
```

```python
xgb_y_pred = xgb.predict(x_test)

actual_xgb_y_pred = np.expm1(xgb_y_pred)

print("RMSE:", np.sqrt(mean_squared_error(y_test_actual, actual_xgb_y_pred)))
print("MAE:", mean_absolute_error(y_test_actual, actual_xgb_y_pred))
print("R2:", r2_score(y_test_actual, actual_xgb_y_pred))
```

```python
scores = cross_val_score(xgb, x, y, cv=5, scoring="r2")
print("CV R2:", scores.mean())
```

```python
import joblib


def preprocess_test(df):

    df = df

    scaler = joblib.load("scaler.pkl")
    neigh_map = joblib.load("neighborhood_map.pkl")
    train_columns = joblib.load("train_columns.pkl")

    df.drop(columns=["Id"], inplace=True, errors="ignore")

    # Same encoding logic
    qual_map = {"Po":1,"Fa":2,"TA":3,"Gd":4,"Ex":5,"unknown":0}
    bsmt_map = {"Unf":0,"LwQ":1,"Rec":2,"BLQ":3,"ALQ":4,"GLQ":5,"unknown":-1}
    func_map = {"Sev":1,"Maj2":2,"Maj1":3,"Mod":4,"Min2":5,"Min1":6,"Typ":7}

    for col in ["ExterCond","BsmtCond"]:
        if col in df.columns:
            df[col] = df[col].map(qual_map)

    if "BsmtFinType1" in df.columns:
        df["BsmtFinType1"] = df["BsmtFinType1"].map(bsmt_map)

    if "BsmtFinType2" in df.columns:
        df["BsmtFinType2"] = df["BsmtFinType2"].map(bsmt_map)

    if "Functional" in df.columns:
        df["Functional"] = df["Functional"].map(func_map)

    # Target encoding using TRAIN map
    df["Neighborhood"] = df["Neighborhood"].map(neigh_map)

    # Rare handling
    freq = df["RoofMatl"].value_counts()
    rare = freq[freq < 10].index
    df["RoofMatl"] = df["RoofMatl"].replace(rare,"Other")

    # Combine exterior
    df["Exterior"] = df["Exterior1st"] + "_" + df["Exterior2nd"]
    df.drop(columns=["Exterior1st","Exterior2nd"], inplace=True)

    df = pd.get_dummies(df, drop_first=True)

    # Align columns with train
    df = df.reindex(columns=train_columns, fill_value=0)

    X_scaled = scaler.transform(df)

    test_df = pd.DataFrame(X_scaled, columns=train_columns)
    test_df.to_csv("processed_test.csv", index=False)

    return test_df
```

```python
test = pd.read_csv(r"C:\potfolio\Advance_house_price_prediction\data\raw\test.csv")

test_ids = test["Id"].copy()   # SAVE ID FIRST

test_df = preprocess_test(test)

test_df.head()
```

```python
stack_pred = stack_model.predict(test_df)
final_pred = np.expm1(stack_pred)

submission = pd.DataFrame({
    "Id": test_ids,
    "SalePrice": final_pred
})

submission.to_csv("stack_submission_6.csv", index=False)
```
