# Phase 1 — Sensor Selection Findings & Next Steps

## Sensor Selection Decisions

### Wind Speed: wind_speed_3 over wind_speed_4
Both sensors measure wind speed at the turbine and show near-identical readings across the full operational period. Statistical validation confirmed a correlation of 0.997 and a mean absolute difference of 0.038 — effectively the same signal. The decisive factor is feature availability: wind_speed_3 provides avg, min, max, and std variants while wind_speed_4 only provides avg. Retaining wind_speed_4 adds no predictive value while reducing feature richness for downstream modeling. wind_speed_3 is selected as the sole wind input for all subsequent analysis.

### Power Output: power_30 over power_29
power_30 measures net export power — the energy actually delivered to the grid after subtracting the turbine's internal consumption (cooling systems, control units, hydraulics). This is the revenue-generating signal and the correct target for generation loss quantification. power_29 measures gross generated power at the generator terminals and does not reflect what the energy company is actually paid for. power_30 contains occasional negative values during low-wind or shutdown states where internal consumption briefly exceeds generation — these are valid operational readings, not data errors. For power curve baseline construction, negative values will be filtered from status_0 (normal operation) rows since active generation should never produce negative net export.

## Next Steps — Power Curve Construction

The next phase is building the expected power curve using status_0 (normal operation) rows only. Wind speed (wind_speed_3_avg) will be plotted against net export power (power_30_avg) to establish the healthy baseline relationship between wind input and power output. This curve defines what the turbine *should* produce at any given wind speed. Deviations from this curve — where actual output falls below expected output for a given wind speed — are generation loss events. Quantifying the area between expected and actual output, translated to MWh and then to revenue using market price, produces the business impact number that anchors Problem 1 of this project.
