normal operation:
  * 0: Normal Operation - The turbine is in normal power production mode 
  * 1: Derated Operation - Derated power generation with a power restriction
  * 2: Idling - Asset is idling and waits to operate again
  * 3: Service - Asset is in service mode / service team is at the site
  * 4: Downtime - Asset is down due a fault or other reasons
  * 5: Other - Other operational states

```python
import plotly.graph_objects as go 
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=file_0.index[12345:22345],
        y=file_0["wind_speed_3_avg"][12345:22345],
        mode="lines+markers",
        name="Speed 3"
    )
)

fig.add_trace(
    go.Scatter(
        x=file_0.index[12345:22345],
        y=file_0["wind_speed_4_avg"][12345:22345],
        mode="lines+markers",
        name="Speed 4"
    )
)

fig.update_layout(
    xaxis_title="Time / Index",
    yaxis_title="Wind Speed Avg",
    title="Time series: Speed 3 vs Speed 4"
)

save_analysis("Time series_Speed 3 vs Speed 4",fig)
fig.show()
```

```python
correlation = file_0["wind_speed_3_avg"].corr(file_0["wind_speed_4_avg"])
mean_diff = (file_0["wind_speed_3_avg"] - file_0["wind_speed_4_avg"]).mean()
max_diff = (file_0["wind_speed_3_avg"] - file_0["wind_speed_4_avg"]).abs().max()

print(f"Correlation: {correlation:.6f}")
print(f"Mean difference: {mean_diff:.6f}")
print(f"Max absolute difference: {max_diff:.6f}")
```
