```python
high_mean_value = ab_nyc.groupby("neighbourhood").agg({
    "price" : [ "median", "mean", "sum"],
    "log_price" : ["median", "mean", "sum"]
}).sort_values(by = ("price","median"), ascending=False)
high_mean_value.head(15)
```

```python
ab_nyc.groupby("neighbourhood_group")["price"].median().sort_values(ascending=False)
```

```python
group_room_price = ab_nyc.groupby("room_type")["price"].median().reset_index().sort_values(by = "price",ascending=False)
fig = px.box(ab_nyc, x="room_type", y ="price", title="Box plot for types of rooms",points="outliers")
fig.update_traces(marker = dict(color = "red", line =dict(color = "black", width = 1)))
fig.update_layout(height = 600)
fig.show()
```

```python
fig = px.box(ab_nyc, x="room_type", y ="log_price", title="Box plot for types of rooms",points="outliers")
fig.update_traces(marker = dict(color = "red", line =dict(color = "black", width = 1)))
fig.update_layout(height = 600)
fig.show()
```

```python
two_way = ab_nyc.groupby(["neighbourhood_group", "room_type"])["price"].median().reset_index().pivot(index="neighbourhood_group", columns="room_type", values="price")
two_way
```

```python
fig = px.imshow(
    two_way,
    text_auto=True,
    color_continuous_scale="RdBu_r",
    title="Median Price by Room Type and Neighbourhood Group"
)
fig.update_layout(height=500, width=1000)
fig.show()
```

**1. Manhattan is the most expensive**

    → For all room types
    → Entire home/apt median ≈ 200+
    → Private rooms also expensive (~100)

**2. Brooklyn is second**

    → Entire home/apt ≈ 150
    → Private rooms ≈ 75

**3. Queens, Bronx, Staten Island are cheaper**

    Queens has mid prices

    Bronx lowest overall

    Staten Island cheap & low supply

**4. Room Type effect is consistent**

    Across all areas:

    Entire home/apt > Private room > Shared room

    This shows that room type has global effect, while neighbourhood has local variation.

```python
df = ab_nyc.dropna(subset=["log_price"])
```

```python
import statsmodels.api as sm 
from statsmodels.formula.api import ols 

model = ols(
    "log_price ~ C(neighbourhood_group) * C(room_type)", data = df
).fit()

anova_table = sm.stats.anova_lm(model , typ =2)
anova_table
```
