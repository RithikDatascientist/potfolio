```python
claims = pd.concat([outpatient, inpatient], axis=0)
claims.head()
```

```python
data = claims.merge(benficiary, on="BeneID", how = "left")
data.head()
```

```python
data.isna().sum()
```

```python
col = data.columns
size  = len(data)
dec= {}

for name in col :
    missing = data[name].isna().sum()/size * 100
    dec[name] = int(missing)

dec
```

```python
result = []

for key, value in dec.items():
    if value < 15:
        result.append(key)
result
```

```python
new_df = data[result[:]]
```

```python
new_df.head()
```

```python
provider_features = new_df.groupby("Provider").agg({
    "ClaimID" : "count",
    'InscClaimAmtReimbursed': ['sum', 'mean', 'std'],
    'BeneID': 'nunique',
    'AttendingPhysician': 'nunique',
    "IPAnnualReimbursementAmt" : ["sum", "mean", "std"],
    "OPAnnualReimbursementAmt" : ["sum", "mean", "std"],
    "IPAnnualDeductibleAmt" : ["sum", "mean", "std"],
    "OPAnnualDeductibleAmt" : ["sum", "mean", "std"],
}).reset_index()

provider_features.columns = ["Provider","ClaimId_count", "ICR_sum","ICR_mean","ICR_std","Ben_nunique","AP_nunique","IPAR_sum","IPAR_mean","IPAR_std",
                             "OPAR_sum","OPAR_mean","OPAR_std","IPDA_sum","IPDA_mean","IPDA_std","OPDA_sum","OPDA_mean","OPDA_std"]
provider_features.head(20)
```

```python
provider_features.shape
```

```python
final_data = provider_features.merge(labels, on = "Provider")
final_data.head()
```

```python
final_data.to_csv(r"C:\potfolio\Health_care_provider_fraud_detection\data\processed_data\final_data.csv")
```
