```python
X = df[["ClaimId_count", "ICR_sum"]].copy()
import numpy as np

X["log_claims"] = np.log1p(X["ClaimId_count"])
X["log_icr"] = np.log1p(X["ICR_sum"])
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X[["log_claims"]], X["log_icr"])
```

```python
X["expected_log_icr"] = lr.predict(X[["log_claims"]])
X["residual"] = X["log_icr"] - X["expected_log_icr"]
```

```python
X["anomaly_score"] = X["residual"].clip(lower=0)
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x = X["expected_log_icr"],
        y = X["residual"],
        mode = 'markers',  
        marker = dict(
            size = 5,
            opacity = 0.6,
            color = X["anomaly_score"],
            colorscale = 'Viridis',
            colorbar = dict(title = "Anomaly Score"),
            line = dict(width = 1, color = 'black')
        ),
        
    )
)

fig.update_layout(
    title = "Relationship between expected_log_icr and residual",
    xaxis_title = "expected_log_icr",
    yaxis_title = "residual",
    template = "plotly_white",
    hovermode = 'closest'
)

fig.show()
```

```python
result = (
    df[["Provider", "PotentialFraud"]]
    .join(X[["anomaly_score", "residual"]])
    .sort_values("anomaly_score", ascending=False)
)
```

```python
result[result["PotentialFraud"] == "Yes"]
```

```python
fig = go.Figure()


fig.add_trace(
    go.Box(
        x = result["PotentialFraud"],
        y = result["anomaly_score"],
        marker = dict(color ="red", line = dict(width = 1,color = "black")),
        boxpoints = 'outliers'
    )
)
fig.update_layout(
    title = "Distribution of Potential Fraud on Anamoaly Score",
    xaxis_title = "Potential Fraud",
    yaxis_title = "Anomaly Score"
)

fig.show()
```
