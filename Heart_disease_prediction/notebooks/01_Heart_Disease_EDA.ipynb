```python
import pandas as pd 
import numpy as np 
import scipy.stats as stats
import plotly.graph_objects as go 
import plotly.express  as px
```

```python
train_df = pd.read_csv(r"C:\potfolio\Heart_disease_prediction\data\raw\train.csv")
train_df.head()
```

```python
train_df.info()
```

```python
class_balance = train_df["Heart Disease"].value_counts().reset_index()
class_balance.columns = ["Type", "value"]
class_balance
```

```python
fig = go.Figure()
fig.add_trace(
    go.Pie(labels=class_balance["Type"], values=class_balance["value"])
)
fig.update_layout(
    title = "Pie chart of class"
)
fig.show()
```

```python
train_df.columns
```

```python
fig = go.Figure()

fig.add_trace(
    go.Histogram(
        x = train_df["Age"],
        marker=dict(
            line = dict(color = "Black", width = 1),
            color = "skyblue"
            ),

        nbinsx=25
    )
)

fig.update_layout(
    title = "Age distribution",
    xaxis_title = "Age scale",
    yaxis_title = "Frequency"
)

fig.show()
```

```python
gender_based_positive = train_df.groupby("Sex")["Heart Disease"].value_counts().reset_index()
gender_based_positive["Sex"] = gender_based_positive["Sex"].replace({0:"Female", 1:"Male"})
gender_based_positive[gender_based_positive["Heart Disease"] == "Presence"]
```

```python
fig = go.Figure()

fig.add_trace(
    go.Bar(
        x = gender_based_positive["Sex"],
        y = gender_based_positive["count"]
    )
)

fig.update_layout(
    title = "Comparison of Presence of Heart Disease between Gender",
    xaxis_title = "Gender",
    yaxis_title = "Count"
)
fig.show()
```
